```python
from copy import deepcopy
import heapq
```

```python
def push(heuristics, depth, path):
    heapq.heappush(pqueue,(heuristics+depth, depth, path))

def pop():
    return heapq.heappop(pqueue)
```

```python
size = int(input())
contents = size**2
seq = [(r, c) for r in range(size) for c in range(size)]
```

```python
def heuristic(state: list):
	dist = 0
	misp = 0
	for idx in range(contents):
		if state[idx] != goal[idx]:
			misp += 1
			m1, n1 = state[idx]//size, state[idx] % size
			m2, n2 = seq[idx]
			dist += abs(m1-m2)+abs(n1-n2)
	return dist+misp
```

```python
def possible(curr):
	a, b = curr//size, curr%size
	temp = [[a, b+1], [a, b-1], [a+1, b], [a-1, b]]
	temp = list(filter(lambda ele: ele[0]>=0 and ele[0]<size and ele[1]>=0 and ele[1]<size,temp))
	temp = [d[0]*size+d[1] for d in temp]
	return temp
```

```python
init = tuple([int(input()) for i in range(contents)])
pqueue = []
visited = set()
push(999, 0, [init])
```

```python
goal = tuple(range(contents))
```

```python
prev = -1
cnt = 1
move = {-size:'DOWN', size:'UP', -1:'RIGHT', 1:'LEFT', 0:'START'}
while True:
    _, depth, path = pop()
    state = path[-1]
    curr = state.index(0)
    visited.add(state)
    swap_index = possible(curr)
    
    if cnt!=1:
        prev = path[-2].index(0)
        swap_index.remove(prev)
    
    if state == goal:
        print(f'stage {cnt},\n Moves:{depth}')
        prev = init.index(0)
        for ind, state in enumerate(path):
            curr = state.index(0)
            print(f'{ind}\t{move[prev-curr]}')
            prev = curr
        break
    
    depth+= 1
    tst = list(state)
    
    for repl in swap_index:
        tst[curr], tst[repl] = tst[repl], tst[curr]
        if tuple(tst) not in visited:
            path.append(tuple(tst))
            push(heuristic(tst), depth, deepcopy(path))
            path.pop(-1)
        tst[curr], tst[repl] = tst[repl], tst[curr]
    
    cnt += 1
```
