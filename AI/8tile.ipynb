```python
from copy import deepcopy
```

```python
def display(init):
	for k, c in enumerate(init):
		print(c,end=" ")
		if k in [2,5,8]:
			print('\n')
	print("-"*10)
```

```python
def heuristic(init: list):
	dist = 0
	misp = 0
	goal = [1, 2, 3, 4, 5, 6, 7, 8, 0]
	for idx in range(9):
		if init[idx] != goal[idx]:
			misp += 1
			m1, n1 = init[idx]//3, init[idx] % 3
			m2, n2 = goal[idx]//3, goal[idx] % 3
			dist += abs(m1-m2)+abs(n1-n2)
	return dist+misp
```

```python
def possible(curr):
	a, b = curr//3, curr%3
	temp = [[a, b+1], [a, b-1], [a+1, b], [a-1, b]]
	temp = list(filter(lambda ele: ele[0]>=0 and ele[0]<3 and ele[1]>=0 and ele[1]<3,temp))
	temp = [d[0]*3+d[1] for d in temp]
	return temp
```

```python
init = (0, 3, 8, 4, 1, 7, 2, 6, 5)
goal = (0, 1, 2, 3, 4, 5, 6, 7, 8)
visited = set()
pqueue = [(0,999,[init])]
```

```python
prev = -1
cnt = 1
while pqueue:
    # print(pqueue)
    depth, _, path = pqueue.pop(0)
    state = path[-1]
    if state == goal:
        print(f'stage {cnt},\n Moves:{len(path)},\n Path:')
        for p in path:
            display(
        del pqueue
        break
    visited.add(state)
    curr = state.index(0)
    swap_index = possible(curr)
    print(f'Stage:{cnt}, curr:{curr}, prev:{prev}, swaps:{swap_index}')
    display(state)
    if cnt!=1:
        swap_index.remove(prev)
    depth+= 1
    tst = list(state)
    for repl in swap_index:
        tst[curr], tst[repl] = tst[repl], tst[curr]
        if tuple(tst) not in visited:
            path.append(tuple(tst))
            pqueue.append((depth, heuristic(tst), deepcopy(path)))
            path.pop(-1)
        tst[curr], tst[repl] = tst[repl], tst[curr]   
    pqueue.sort(key = lambda x:x[0]+x[1])
    prev = curr
    cnt += 1
    
```
